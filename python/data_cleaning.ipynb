```python
import pandas as pd
```

```python
#Read each of the sheet into its own table
bookings = pd.read_excel("Customer_Travel_Booking_Analytics_Dataset.xlsx", sheet_name="Bookings")
customers = pd.read_excel("Customer_Travel_Booking_Analytics_Dataset.xlsx", sheet_name="Customers")
consultants = pd.read_excel("Customer_Travel_Booking_Analytics_Dataset.xlsx", sheet_name="Consultants")
destinations = pd.read_excel("Customer_Travel_Booking_Analytics_Dataset.xlsx", sheet_name="Destinations")
```

```python
#check for missing values
print("Missing values in Bookings:")
print(bookings.isna().sum())

print("\nMissing values in Customers:")
print(customers.isna().sum())

print("\nMissing values in Consultants:")
print(consultants.isna().sum())

print("\nMissing values in Destinations:")
print(destinations.isna().sum())
```

```python
#check for duplicate rows

print("Duplicate rows in Bookings:", bookings.duplicated().sum())
print("Duplicate rows in Customers:", customers.duplicated().sum())
print("Duplicate rows in Consultants:", consultants.duplicated().sum())
print("Duplicate rows in Destinations:", destinations.duplicated().sum())
```

```python
#standardize date formats

bookings["Booking Date"] = pd.to_datetime(bookings["Booking Date"])
bookings["Travel Date"] = pd.to_datetime(bookings["Travel Date"])

print(bookings[["Booking Date", "Travel Date"]].dtypes)
```

```python
bookings = bookings.drop_duplicates().dropna()
customers = customers.drop_duplicates().dropna()
consultants = consultants.drop_duplicates().dropna()
destinations = destinations.drop_duplicates().dropna()
```

```python
#Save the cleaned sheets as CSV

bookings.to_csv("cleaned_bookings.csv", index=False)
customers.to_csv("cleaned_customers.csv", index=False)
consultants.to_csv("cleaned_consultants.csv", index=False)
destinations.to_csv("cleaned_destinations.csv", index=False)

print("Done!")
```
